```python
import matplotlib.pyplot as plt
import numpy as np

# Data
low = np.array([
    [6, 2, 1],
    [8, 4, 2]
])

mid = np.array([
    [8, 4, 2],
    [10, 6, 4]
])

high = np.array([
    [10, 6, 4],
    [12, 8, 6]
])

datasets = [low, mid, high]

x_labels = ["Small (<=2)", "Average (3-4)", "Large (>=5)"]
y_labels = ["Male children", "Female children"]

titles = [
    "Wealth: Low (<$500)",
    "Wealth: Mid (<$775)",
    "Wealth: High (>$775)"
]

# Figure
fig, axes = plt.subplots(1, 3, figsize=(18, 5), constrained_layout=True)

vmin = 0
vmax = 12

for ax, data, title in zip(axes, datasets, titles):

    im = ax.imshow(data, cmap="viridis", aspect="auto", vmin=vmin, vmax=vmax)

    ax.set_xticks(np.arange(len(x_labels)))
    ax.set_yticks(np.arange(len(y_labels)))

    ax.set_xticklabels(x_labels, fontsize=11)
    ax.set_yticklabels(y_labels, fontsize=11)

    ax.set_title(title, fontsize=16, pad=10)

    # Annotate with hours only
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            ax.text(
                j, i,
                f"{data[i, j]}",
                ha="center",
                va="center",
                color="white",
                fontsize=12,
                fontweight="bold"
            )

# Shared colorbar
cbar = fig.colorbar(im, ax=axes.ravel().tolist(), shrink=0.85, pad=0.02)
cbar.set_label("Hours/week", rotation=270, labelpad=20, fontsize=12)

# Title
fig.suptitle(
    "Parental time per child by Wealth × Family Size × Gender (hours/week)\n",
    fontsize=20,
    y=1.05
)

# Footnote
fig.text(
    0.5,
    -0.03,
    "Made by author based on interview results",
    ha="center",
    fontsize=14
)

plt.show()

# Save figures
fig.savefig("parental_time_heatmap_clean.png", dpi=300, bbox_inches="tight")
fig.savefig("parental_time_heatmap_clean.pdf", dpi=300, bbox_inches="tight")


fig.savefig(r"C:\RES\parental_time_heatmap.png", dpi=300, bbox_inches="tight")
fig.savefig(r"C:\RES\parental_time_heatmap.pdf", dpi=300, bbox_inches="tight")


```
