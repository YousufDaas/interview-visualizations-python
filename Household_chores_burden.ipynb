```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Data
# -----------------------------
data = np.array([
    [6, 8, 10],    # Male children
    [10, 14, 18]   # Female children
])

x_labels = ["Small\n(<=2 school-aged)", "Average\n(3-4)", "Large\n(>=5)"]
y_labels = ["Male children", "Female children"]

# -----------------------------
# Figure
# -----------------------------
fig, ax = plt.subplots(figsize=(10, 6))

im = ax.imshow(data, cmap="viridis", vmin=6, vmax=18)

# Axis ticks and labels
ax.set_xticks(np.arange(len(x_labels)))
ax.set_yticks(np.arange(len(y_labels)))
ax.set_xticklabels(x_labels, fontsize=12)
ax.set_yticklabels(y_labels, fontsize=12)

# Axis titles
ax.set_xlabel("Family size", fontsize=14)
ax.set_ylabel("Gender", fontsize=14)

# Main title
ax.set_title(
    "Household chores burden (hours/week) — by Gender × Family Size",
    fontsize=16,
    pad=12
)

# -----------------------------
# Annotate cells (WHITE + BOLD)
# -----------------------------
for i in range(data.shape[0]):
    for j in range(data.shape[1]):
        ax.text(
            j, i,
            f"{data[i, j]}",
            ha="center",
            va="center",
            color="white",
            fontsize=14,
            fontweight="bold"
        )

# Gridlines
ax.set_xticks(np.arange(-.5, data.shape[1], 1), minor=True)
ax.set_yticks(np.arange(-.5, data.shape[0], 1), minor=True)
ax.grid(which="minor", color="#b7b7b7", linestyle="--", linewidth=1, alpha=0.6)
ax.tick_params(which="minor", bottom=False, left=False)

# Colorbar
cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.03)
cbar.set_label("Hours per week", rotation=270, labelpad=18, fontsize=14)

# Source note
fig.text(
    0.5,
    0.03,
    "Source: Made by author based on interviews results",
    ha="center",
    fontsize=16,
    family="serif"
)

plt.tight_layout(rect=[0, 0.06, 1, 1])
plt.show()

# Save figures
fig.savefig("household_chores_burden_heatmap.png", dpi=300, bbox_inches="tight")
fig.savefig("household_chores_burden_heatmap.pdf", dpi=300, bbox_inches="tight")
```
